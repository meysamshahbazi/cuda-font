```python
from PIL import Image, ImageFont, ImageDraw
import numpy as np
```

```python
im_l = Image.new('L', (4749, 60), ( 0))

# text = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
text = ""
for i in range(32, 255):
    text = text + chr(i)

font = ImageFont.truetype(font="/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf", size=40)
drawer = ImageDraw.Draw(im_l)

fill_color = (255, 0, 0)
stroke_color = (0, 0, 255)

drawer.text((0, 0), text, font=font, fill=255, stroke_width=1, stroke_fill=0)
```
