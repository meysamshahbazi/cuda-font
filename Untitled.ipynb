```python

font_map_i = Image.new('RGB', (512, 512), (0,0,0))


# font = ImageFont.truetype(font="/usr/share/fonts/truetype/liberation/LiberationSans-Regular.ttf", size=40)
# font = ImageFont.truetype(font="/home/meysam/Downloads/dripicons-v2.ttf", size=20)
# font = ImageFont.truetype(font="/home/meysam/Downloads/Action-Man/Action_Man.ttf", size=40)
# font = ImageFont.truetype(font="/home/meysam/Downloads/FontAwesome.ttf", size=20)

# font_name = "/home/meysam/Downloads/dripicons-v2.ttf"
# font_name = "/home/meysam/Downloads/beautiful_6/Beautiful.ttf"
# font_name = "/home/meysam/Downloads/sylvia_angel/Sylvia Angel.otf"
# font_name = "/home/meysam/Downloads/harvest_yard/Harvest Yard.otf"
# font_name  = "/home/meysam/Downloads/FontAwesome.ttf"
# font_name  =  "/home/meysam/Downloads/Action-Man/Action_Man.ttf"
# font_name  =  "/home/meysam/Downloads/Action-Man/Action_Man_Extended.ttf"
# font_name  =  "/home/meysam/Downloads/ready_for_fall/Ready For Fall.ttf"
# font_name  =  "/home/meysam/Downloads/super_bakery/Super Bakery.ttf"
# font_name  = "/home/meysam/Downloads/hug_me_tight/Hug Me Tight - TTF.ttf"
# font_name  = "/home/meysam/Downloads/simple_handmade_2/SimpleHandmade.ttf"

font_name = "/home/meysam/Downloads/mouldy_cheese/Mouldy Cheese.ttf"
# font_name = "/home/meysam/Downloads/sparky_stones/Sparky Stones.ttf"
font = ImageFont.truetype(font=font_name, size=30)


drawer = ImageDraw.Draw(font_map_i)

text = "" 

for i in range(32, 255):
    text = text + chr(i)

# text = "" + chr(171) 
# text = "\xB0"

fill = (0,255,0)
stroke_width = 3
stroke_fill = (255,0,0)
GlyphInfo = dict()

x1, y1, x2, y2 = font.getbbox(text, stroke_width=stroke_width, anchor = "la")
max_height = y2 - y1

min_x_draw = stroke_width

x_draw = min_x_draw
y_draw = stroke_width

x1, y1, x2, y2 = font.getbbox("!", stroke_width=stroke_width, anchor = "la")
min_x_adv = x2 - x1 
for l in text:
    GlyphInfo[l] = dict()
    x1, y1, x2, y2 = font.getbbox(l, stroke_width=stroke_width, anchor = "la")
    if x_draw + x2 - x1 > 512:
        x_draw = min_x_draw
        y_draw += max_height
        
    GlyphInfo[l]["x"] = x_draw + x1
    GlyphInfo[l]["y"] = y_draw + y1
    GlyphInfo[l]["width"] = x2 - x1
    GlyphInfo[l]["height"] = y2 - y1
    GlyphInfo[l]["xAdvance"] = x2 - x1
    if x2 - x2 == 2*stroke_width:
       GlyphInfo[l]["xAdvance"] = min_x_adv
        
    GlyphInfo[l]["xOffset"] = -x2
    GlyphInfo[l]["yOffset"] = -y2 + y1
    drawer.text((x_draw, y_draw), l, font=font, anchor = "la", 
                fill=fill, stroke_width=stroke_width, stroke_fill=stroke_fill)
    x_draw += x2 - x1 + min_x_draw

font_map_i
```

```python
with open("GlyphInfo.txt", "w+") as f:
    for k in GlyphInfo.keys():
        line = ""
        line  += str(GlyphInfo[k]["x"]) + ","
        line  += str(GlyphInfo[k]["y"]) + "," 
        line  += str(GlyphInfo[k]["width"]) + "," 
        line  += str(GlyphInfo[k]["height"]) + ","
        line  += str(GlyphInfo[k]["xAdvance"]) + ","
        line  += str(GlyphInfo[k]["xOffset"]) + ","
        line  += str(GlyphInfo[k]["yOffset"]) + ",\n"
        f.writelines(line)
```
